```python
import numpy as np
import pandas as pd
import yfinance as yf
```

```python
data_inicio = '2012-01-01'
data_fim = '2026-03-01'
data_backtest = "2022-07-01" 
```

```python
ativos = ['ABCB4', 'ALOS3', 'BRSR6', 'BBSE3', 'BRBI11', 'BBDC3', 'BBDC4', 'BRAP4', 'BBAS3', 'AGRO3', 'CXSE3', 'CMIG4', 'CSMG3', 'CPFE3', 'CMIN3', 'CURY3', 'DIRR3', 'EGIE3', 'EVEN3', 'EZTC3', 'FESA4', 'FLRY3', 'GRND3', 'RANI3', 'ISAE4', 'ITSA4', 'ITUB3', 'ITUB4', 'JHSF3', 'KEPL3', 'KLBN11', 'LAVV3', 'LOGG3', 'POMO4', 'MBRF3', 'LEVE3', 'ODPV3', 'PGMN3', 'PETR3', 'PETR4', 'RECV3', 'SAPR11', 'SLCE3', 'SYNE3', 'TAEE11', 'TGMA3', 'TIMS3', 'UNIP6', 'VALE3', 'VLID3', 'VBBR3', 'VULC3']

ativos_sa = []
for ativo in ativos:
    ativos_sa.append(ativo + '.SA')

n_ativos   = len(ativos)
carteira = sorted(ativos)
print(ativos_sa)
```

```python
dados = yf.download(ativos_sa, start=data_inicio, end=data_fim)['Close']

retornos = dados.pct_change().dropna()

#pesos por blacklitterman
#pesos = np.array([0.25, 0.25, 0.25, 0.25])
pesos = np.full(n_ativos, 1/n_ativos)

# Calcular a série de retornos históricos do portfólio
retorno_portfolio = retornos.dot(pesos)

# 4. Calcular o VaR e o CVaR Histórico
nivel_confianca = 0.95
alpha = 1 - nivel_confianca # 0.05 para 95% de confiança

# O VaR histórico é o percentil (alpha) da distribuição dos retornos
# Ex: O limite que separa os 5% piores dias dos 95% "normais"
var_95 = retorno_portfolio.quantile(alpha)

# O CVaR é a média condicional: a média de todos os retornos que são menores (piores) que o VaR
cvar_95 = retorno_portfolio[retorno_portfolio <= var_95].mean()

# Multiplicamos por -1 apenas para apresentar a perda como um número positivo (convenção de mercado)
print(f"VaR (95%): {-var_95:.4%}")
print(f"CVaR (95%): {-cvar_95:.4%}")
```
